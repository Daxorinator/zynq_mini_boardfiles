# ZYNQ MINI with zynqmini
Load a PL design, talk to its registers, blink the LEDs. Run as root on the board (the Jupyter service does). The designs are built with the open flow (`openflow/pl`) and installed in `/usr/share/zynqmini/bitstreams`.

```python
from zynqmini import Overlay
ol = Overlay('zm_base')   # loads zm_base.bin, sets FCLK0 to 100 MHz, maps the registers
ol
```

```python
print(hex(ol.ID), 'MMCM locked:', ol.MMCM_LOCKED)
print('keys (0 = pressed):', bin(ol.KEYS))
```

## LEDs
The four PL LEDs are bits 3:0 of `LEDS`.

```python
import time
for i in range(12):
    ol.LEDS = 1 << (i % 4)
    time.sleep(0.15)
ol.LEDS = 0
```

## Registers
`ol.registers` lists every register with its offset and access; `ol.TIMER_CTRL`, `ol.METER_100` etc. are attributes.

```python
for r in ol.registers.values():
    print(f'{r.name:14s} {r.offset:#06x} {r.access:3s} {r.description}')
```
