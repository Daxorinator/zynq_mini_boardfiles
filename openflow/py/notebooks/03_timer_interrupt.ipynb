# PL timer and interrupt
A PL timer counts aclk cycles and raises IRQ_F2P[0], seen by Linux as a UIO event.

```python
import time
from zynqmini import Overlay
ol = Overlay('zm_base')
irq = ol.irq()
ol.TIMER_PERIOD = 10_000_000          # 0.1 s at 100 MHz
ol.TIMER_STATUS = 1
ol.TIMER_CTRL = 0b11                  # enable + interrupt enable
t0 = time.time(); n = 0
while n < 10:
    irq.rearm(); ol.TIMER_STATUS = 1
    if irq.wait(1.0) is None: break
    n += 1
ol.TIMER_CTRL = 0
irq.close()
print(n, 'interrupts in', round(time.time() - t0, 3), 's')
```
