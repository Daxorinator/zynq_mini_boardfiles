# PL clocks
The PS gives the PL up to four FCLKs. `zm_base` has frequency meters (cycles counted in a window of 2^20 FCLK0 cycles) to check them.

```python
from zynqmini import Overlay, set_fclk, fclk_mhz, pll_mhz
import time
ol = Overlay('zm_base')
print('IO PLL:', pll_mhz(2), 'MHz')
```

```python
def meter(name, mhz_ref=100.0):
    time.sleep(0.05)
    return getattr(ol, name) / 2**20 * mhz_ref   # window = 2^20 cycles of FCLK0 (100 MHz)
for f in (50, 75, 100, 125, 150, 200):
    got = set_fclk(1, f)
    time.sleep(0.05)
    print(f'FCLK1 requested {f:5.1f} MHz -> set {got:8.3f} MHz, measured {meter("METER_FCLK1"):8.3f} MHz')
```

```python
print('PL oscillator (50 MHz):', round(meter('METER_50'), 3), 'MHz')
print('MMCM outputs:', [round(meter(n), 2) for n in ('METER_100', 'METER_150', 'METER_200')], 'MHz')
```
